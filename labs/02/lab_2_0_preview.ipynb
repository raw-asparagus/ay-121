```python
from pathlib import Path

import pandas as pd

from ugradiolab import Spectrum

from utils.lab_2_0_preview_plot import plot_dataset_pair
from utils.preview_data import build_preview_rows, choose_zero_files
```

```python
DATA_ROOT = Path('data')
LO_FREQS_MHZ = (1420, 1421)
RUN_INDEX = 0

DATASETS = [
    {'label': 'standard', 'raw_dir': DATA_ROOT / 'standard'},
    {'label': 'cygnus-x', 'raw_dir': DATA_ROOT / 'cygnus-x'},
    {'label': 'human', 'raw_dir': DATA_ROOT / 'human'},
    {'label': 'SGP', 'raw_dir': DATA_ROOT / 'SGP'},
    {'label': 'cold_ref', 'raw_dir': DATA_ROOT / 'cold_ref'},
]
```

```python
# Preview-data helpers are imported from utils.preview_data.
```

```python
pairs_by_dataset: dict[str, dict[int, Spectrum]] = {}

for dataset in DATASETS:
    label = dataset['label']
    raw_dir = dataset['raw_dir']

    print()
    print('=' * 80)
    print(f'[{label}]')

    raw_paths_by_lo = choose_zero_files(raw_dir, lo_freqs_mhz=LO_FREQS_MHZ, run_index=RUN_INDEX)
    pairs_by_dataset[label], rows = build_preview_rows(raw_paths_by_lo)
    display(pd.DataFrame(rows).set_index('LO (MHz)'))
```

```python
savgol = dict(method='savgol', window_length=257, polyorder=3)

for dataset in DATASETS:
    label = dataset['label']
    print()
    print('=' * 80)
    print(f'[{label}] plotting stage')

    spec_1420 = pairs_by_dataset[label][1420]
    spec_1421 = pairs_by_dataset[label][1421]
    plot_dataset_pair(label, "raw", spec_1420, spec_1421, savgol)
```
